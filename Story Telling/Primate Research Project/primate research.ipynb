Can books change cultures?

The 1975 publication of the book Animal Liberation by Peter Singer is often credited with sparking an animal rights movement. Did it? Let's look at word usage data to explore this claim? We will focus on primates used in biomedical research.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('animal-word-trends-primate-research.csv')
df.head()
````

````python
def plot_word_trend( animal ):
    trend = df.query('word == @animal')
    plt.figure(figsize=(5, 3))
    plt.plot(trend['year'], trend['frequency'], label=animal)
    plt.ylabel('Frequency per million')
    plt.title(f'Word usage for "{animal}" and year Animal Liberation book published')
    plt.axvline(1975, color='orange', ls='--')
````

Explore word trends for primates used in research.

````python
plot_word_trend('rhesus')
# try primate, marmoset, baboon, rhesus
````

### How animal research changed

Let's look at another dataset to see what was happening with the research.

````python
counts = pd.read_csv('number-of-research-animals-per-year.csv')
counts.head(2)
````

````python
def plot_animal_research( animal ):
	plt.figure(figsize=(5, 3))
	plt.plot(counts['year'], counts[animal])
	plt.axvline(1975, label='Animal Liberation book', color='orange', ls='--')
	plt.ylabel('Number of animals')
	plt.title(f'Number of {animal} used in research per year')
	plt.legend()
	plt.show()

plot_animal_research('primates')
````

# **1. Word Usage Trends in Relation to Animal Liberation (1975)**

The word usage data reveals a striking correlation between the publication of Peter Singer’s Animal Liberation and the discourse surrounding primates.

*   **Synchronized Peaks:** Every primate-related term analyzed (baboon, marmoset, primate, and rhesus) reached its historical peak in frequency between 1975 and 1976.

*   **Post-1975 Decline:** After this sharp peak, the frequency of these words in literature began a gradual decline. This suggests that 1975 was a "flashpoint" year where public and academic discussion of primate research was at its absolute highest.

# **2. Trends in the Number of Primates Used in Research**

The data for actual animal usage (1973–2019) shows a different story than the word usage trends:

*   **General Increase:** While word usage declined after 1975, the actual number of primates used in research has trended upward over time.

*   **The "Primate Exception":** Primates are the only animal group in the dataset that has seen a long-term increase in usage. While the use of dogs, cats, rabbits, guinea pigs, and hamsters has declined significantly since the 1970s, primate usage has grown at an average rate of approximately 456 animals per year.

*  **Initial Dip:** There was a notable dip in primate research numbers in 1975 (36,202 animals), but the numbers recovered quickly, reaching over 70,000 in recent years.

# **3. Potential Explanations for Word Usage Trends**

*   **The Discourse Shock:** The peak in 1975-1976 likely reflects the "shock" to the cultural and scientific system caused by the animal rights movement. The book brought primate research into the mainstream spotlight, leading to a surge in books, articles, and debates.

*   **Normalization or Shift:** The subsequent decline in word frequency could indicate that the topic became "normalized" or that the specific vocabulary of the 1970s animal rights debate evolved into different terms not captured in this specific word list.

*   **Scientific Resilience:** The fact that primate numbers increased while word usage (and other animal usage) decreased suggests that primates occupy a unique niche. As ethical pressure grew to reduce animal testing, researchers likely successfully reduced the use of "simpler" animals (like dogs or rabbits) but found primates to be irreplaceable for complex biomedical research (e.g., neuroscience, vaccines, and immunology).

# **4. Conclusions on the Claim that Animal Liberation Sparked a Movement**

The data supports the claim that the book sparked an animal rights movement, but with nuances:

*   **Success in Awareness:** The word usage data provides clear evidence of a massive cultural shift in discourse starting exactly in 1975.

*   **Success in Substitution:** The sharp decline in the use of dogs, cats, and rabbits suggests the movement was highly effective at reducing the research footprint for many species.

*   **The Primate Paradox:** The movement did not succeed in reducing the number of primates used in research. In fact, primate research has intensified. This suggests that while the movement changed how we talk about animals and which animals we use, the scientific necessity of primates has, thus far, outweighed the ethical movement's goal of total abolition.

**In summary**, Animal Liberation successfully changed the cultural conversation and reduced the use of many laboratory animals, but primate research remains a growing and persistent part of the scientific landscape.
