# Silent Spring

When the pesticide DDT was made available to the public in 1944, it was seen as a miracle chemical in the battle against insect pests. How did the worldwide availability of DDT coincide with word trends for insect words?

```python
# FOR GOOGLE COLAB ONLY.
# Uncomment and run the code below. A dialog will appear to upload files.
# Upload 'animal-word-trends-silent-spring.csv' and 'sparrowhawk-eggshell-data.csv'.

from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('animal-word-trends-silent-spring.csv')
df.head()
```

```python
def plot_insect_word_trend( animal ):
    trend = df.query('word == @animal')
    plt.figure(figsize=(5, 3))
    plt.plot(trend['year'], trend['frequency'], label=animal)
    plt.ylabel('Frequency per million')
    plt.title(f'Word usage for "{animal}" and the year DDT was made available')
    plt.axvline(1944, color='orange', ls='--')
```

Explore word trends for insects relative to 1944.

```python
plot_insect_word_trend('louse')
# try flea, mosquito, fly, bedbug, louse
```

#### Impact of DDT

Early marketing for DDT claimed it only killed insect pests and did not affect other animals. The following python script might help shed light on this claim.

```python
egg = pd.read_csv('sparrowhawk-eggshell-data.csv')
plt.figure(figsize=(5, 3))
plt.plot(egg['year'],egg['avg_thickness'], color='C5')
plt.xlabel('Year')
plt.ylabel('Thickness Index')
plt.title('Average Eggshell Thickness')
plt.axvline( 1944, color='magenta', ls='--', label='DDT made available to public' )
plt.show()
```

#### Rachel Carson

In 1962, Rachel Carson published the book Silent Spring, which showed that the toxins killing insects were also harming the rest of the food web. This book began to raise public awareness of DDT's toxicity, which eventually lead to bans on DDT.

In particular, birds of prey such as peregrine falcons, and songbirds such as robins, were highly affected by widespread use of DDT.

Use the function below to explore the word usage levels for these birds relative to the 1962 book.


```python
def plot_bird_word_trend( animal ):
    trend = df.query('word == @animal')
    plt.figure(figsize=(5, 3))
    plt.plot(trend['year'], trend['frequency'], label=animal)
    plt.ylabel('Frequency per million')
    plt.title(f'Word usage for "{animal}" and the year Silent Spring published')
    plt.axvline(1962, color='orange', ls='--')
```

```python
plot_bird_word_trend('condor')
# try robin, hawk, peregrine, osprey, sparrowhawk, condor
```

## **Solutions**

## **1. Word Usage Patterns: Birds vs. Insects**

  Based on the analysis of the animal-word-trends-silent-spring.csv data (visualized in the plots below), we can observe distinct usage patterns for these two groups:

    Insects (bedbug, flea, fly, louse, mosquito):

  **Pattern:** There is a general downward trend in the frequency of insect words throughout the 20th century.

  **Specifics:** The decline is most noticeable after the 1940s. This corresponds to the period when synthetic pesticides like DDT were introduced and widely used, potentially reducing both the actual number of insects and the public's preoccupation with them as pests.

    Birds (condor, hawk, osprey, peregrine, robin, sparrowhawk):

  **Pattern:** Bird word usage shows a "U-shaped" or recovery pattern.

  **Specifics:** There is a noticeable dip or stagnation in the frequency of bird-related words between 1944 and the early 1960s. Following the publication of Silent Spring in 1962, there is a sharp and sustained increase in the frequency of these words, particularly for the robin and birds of prey.

### **The Difference:**
  While insect mentions generally decreased as they were "conquered" by technology, bird mentions suffered a decline during the height of pesticide use but experienced a massive cultural resurgence after the environmental movement began in the 1960s.

# **2. Potential Explanations for These Differences**

The differences in these patterns can be explained by the intersection of chemistry, biology, and environmental history:

*   **The "Miracle" of DDT (1944):** When DDT was released to the public in 1944, it was incredibly effective at killing insects. This explains the continued decline in insect-related words; they were no longer an "unbeatable" problem, and their populations were being suppressed.

*   **Biomagnification and Eggshell Thinning:** While DDT killed insects, it also accumulated in the food chain (biomagnification). Birds of prey (like hawks, ospreys, and peregrines) are at the top of the food chain and received the highest doses.

*   **Evidence from Data:** Looking at the sparrowhawk-eggshell-data.csv, we see a dramatic drop in average eggshell thickness starting almost immediately after 1944.

**Result:** Thinner shells meant eggs would break in the nest, leading to a collapse in bird populations. This biological decline is reflected in the lower frequency of bird names in literature and news during the 1950s.

*   **The "Silent Spring" Effect (1962):** Rachel Carson’s book Silent Spring (1962) warned that the loss of birds would lead to a world where no birds sing in the spring.

*   **Impact:** The book sparked the modern environmental movement. The subsequent ban on DDT (1972 in the US) allowed bird populations to recover. The sharp rise in bird word frequencies after 1962 reflects this newfound public awareness, conservation efforts, and the eventual return of these species to the landscape.
